```python
from typing import TypedDict
class AgentState(TypedDict):
    message:str
    event:str
    response:str
```

```python
def decision_node(state:AgentState):
    if state["event"]=="USER_REGISTERED":
        return "welcome"
    elif state["event"]=="WALLET_CREDIT":
        return "wallet"
    elif state["event"]=="TRANSACTION_WALLET":
        return "transaction"
```

```python
#                     Node 1 
def analyze_event(state:AgentState):
    return {
        "event":state["event"]
    } 
```

```python
def welcome_node(state:AgentState):
    return {
        "response":"Welcome to multiwallet!"
    }

def wallet_node(state:AgentState):
    return {
        "response":"your Wallet Successfully Created"
    }

def transaction_node(state:AgentState):
    return {
        "response":"your Transaction successful"
    }
```

```python
from langgraph.graph import StateGraph,START,END
graph=StateGraph(AgentState)
```

```python
graph.add_node("analyze",analyze_event)
graph.add_node("welcome",welcome_node)
graph.add_node("wallet",wallet_node)
graph.add_node("transaction",transaction_node)
```

```python
graph.add_edge(START,"analyze")
graph.add_conditional_edges(
    "analyze",
    decision_node,
    {
        "welcome":"welcome",
        "wallet":"wallet",
        "transaction":"transaction"
    }
)
graph.add_edge("welcome",END)
graph.add_edge("wallet",END)  
graph.add_edge("transaction",END)  

app=graph.compile()
```

```python
result = app.invoke({
    "message": "Wallet transfer ",
    "event": "WALLET_CREDIT",
    "response": ""
})

print(result)
```
