```python
from mcp.server import MCPServer
mcp=MCPServer("Github_Code_Review")
#if __name__== "__main__":
 #   mcp.run()
```

```python
import threading

def start_server():
    mcp.run(
        transport="streamable-http",
        host="127.0.0.1",
        port=8000
    )

thread = threading.Thread(target=start_server, daemon=True)
thread.start()

print("MCP Server started")
```

```python
@mcp.tool()
def get_file_content(filename:str)->str:
    """ Return the content of the project file """
    files={
        "main.py":"print('hello world')",
        "auth.py":"def login():pass",
        "databases.py":"import psycopg2"
    }

    if filename not in files:
        return f"File '{filename}' not found"
    return files[filename] 
```

```python
from pydantic import BaseModel

class SearchInput(BaseModel):
    filename:str
    keyword:str
```

```python
@mcp.tool()
def search_file(state:SearchInput)->str:
    """ Search for a keyword inside a project file. """

    files2={
        "main.py":"print('hello world')\n print('python')",
        "auth.py":"def login(): pass \n def logout(): pass",
        "databases.py":"import pycopg2\nconnect_databases()"
    } 
    if filename not in files2:
        return f"File '{filename} not found"

    content=files2[filename]

    if keyword in content:
        return f"'{keyword}'  found in {filename}"

    return f"'{keyword}' not found in {filename} "
```

```python
@mcp.tool()
def list_project_files()->list[str]:
    """ Return the files available in the project. """ 
    return [
        "main.py",
        "auth.py",
        "database.py"
        "requirements.txt"
    ]
```

```python
from mcp import ClientSession
from mcp.client.streamable_http import streamable_http_client

async def test_mcp_server():
    async with streamable_http_client("http://127.0.0.1:8000/mcp") as (read_stream,write_stream):
        async with ClientSession(read_stream,write_stream)as session:
            await session.initialize()
            result=await session.call_tool(
                "list_project_files",
                {}
            )
            print(result)
            tools=await session.list_tools()
            print("Avalable tools")
            for tool in tools.tools:
                print(tool.name)
```

```python
await test_mcp_server()
```
