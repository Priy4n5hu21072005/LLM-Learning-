```python
from mcp.server import MCPServer 
mcp=MCPServer("CODE_REVIEW_SERVER")

@mcp.tool()
def get_review_context(context:str)->str:
    """Return the available code review on the basis of context""" 
    return f"Focus on the {context}"

@mcp.tool()
def list_project_files() -> str:
    """Return the files available in the project."""
    return "main.py, auth.py, database.py, requirements.txt"


#if __name__=="__main__":
 #   mcp.run()
```

```python
import threading

def start_server():
    mcp.run(
        transport="streamable-http",
        host="127.0.0.1",
        port=8000
    )

thread = threading.Thread(target=start_server, daemon=True)
thread.start()

print("MCP Server started at http://127.0.0.1:8000")
```

```python
@mcp.resource("project://info")
def project_info() -> str:
    return """
    AI Code Review Agent

    Purpose:
    Review GitHub repositories based on user-selected context.

    Review contexts:
    - Security
    - Performance
    - Code Quality
    - Architecture
    - Bug Detection
    """
```

```python
@mcp.prompt()
def security_review()->str:
    return """
    Review the provided code with a security-focused approach.
    
    check for:
    -Authentication issues
    -Authorization issues
    -Hardcoded secrets
    -Input validation problems
    -unsafe data handling
     """
```

```python
from mcp import ClientSession
from mcp.client.streamable_http import streamable_http_client

async def test_mcp_server():
    async with streamable_http_client("http://127.0.0.1:8000/mcp") as (read_stream,write_stream):
        async with ClientSession(read_stream,write_stream)as session :
            await session.initialize()
            tools=await session.list_tools()
            resources=await session.list_resources()
            print("avaliable resources")
            for resource in resources.resources:
                print(resource.uri)
            print("availablE tools")
            for tool in tools.tools:
                print(tool.name)
            prompts=await session.list_prompts()
            print("Available prompts:")
            for prompt in prompts.prompts:
                print(prompt.name)
            result=await session.call_tool("get_review_context",{"context":"security"})
            resource_data=await session.read_resource("project://info")
            prompt_result= await session.get_prompt("security_review")
            print("resource data:")
            print(resource_data)
            print("result:")
            print(result)
            print("GET prompt:")
            print(prompt_result)
```

```python
await test_mcp_server()
```
