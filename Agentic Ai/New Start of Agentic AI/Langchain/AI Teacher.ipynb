```python
from langchain_core.prompts import PromptTemplate
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
GOOGLE_GEN_AI_API_KEY= "YOUR_GOOGLE_API_KEY"
```

```python
AI = ChatGoogleGenerativeAI(google_api_key=GOOGLE_GEN_AI_API_KEY,model="gemini-3.6-flash")
template = PromptTemplate(
    input_variables=["topic"],
    template=""" You Are the Very Polite and Friendly Teacher who teach the given {topic} in hinglish and try to explain the topic to the 5 years old child""" 
)
```

```python
topic = input("What do you want to learn !!!")
prompt = template.format(topic = topic)
response = AI.invoke(prompt)
response.content
```
