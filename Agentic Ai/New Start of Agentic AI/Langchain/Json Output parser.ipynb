```python
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser,JsonOutputParser
```

```python
GOOGLE_GEN_AI_API_KEY = "YOUR_GOOGLE_API_KEY"
LLM = ChatGoogleGenerativeAI(google_api_key=GOOGLE_GEN_AI_API_KEY,model="gemini-3.5-flash")
parser = JsonOutputParser()
```

```python
prompt = PromptTemplate(
    input_variables=["movie_name"],
    template="""
    Provide information about the movie {movie_name}.
    Return the response strictly in JSON format with keys: 
    "title", "genre", "rating", "summary".
    
    {format_instructions}
    """,
    partial_variables={"format_instructions": parser.get_format_instructions()}
)
```

```python
chain = prompt|LLM|parser
```

```python
result = chain.invoke(
    {
        "movie_name":"Oddesy"
    }
)
result
```
