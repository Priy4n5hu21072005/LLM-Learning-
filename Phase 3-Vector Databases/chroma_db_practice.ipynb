```python
import chromadb
client=chromadb.Client()
client
```

```python
collections=client.create_collection("my_document")
collections
```

```python
documents=[
    "I love programming.",
    "I really enjoy coding.",
    "Programming is my favorite thing."
]
ids=["doc1","doc2","doc3"]
print(documents)
ids
```

```python
from sentence_transformers import SentenceTransformer
model=SentenceTransformer("all-MiniLm-L6-v2")
```

```python
documents_embedding=model.encode(documents)
documents_embedding
```

```python
collections.add(
    ids=ids,
    documents=documents,
    embeddings=documents_embedding.tolist()
)
collections.count()
```

```python
collections.get()
```

```python
client.delete_collection(name="my_document")
```

```python
collections=client.create_collection(name="my_document")
collections
```

```python
metadatas=[
    {
        "topic":"Machine Learning",
        "language":"python"
    },
    {
        "topic":"Backend",
        "language":"java"
    },
    {
        "topic":"sports",
        "language":"None"
    }
]
```

```python
collections.add(
    ids=ids,
    documents=documents,
    embeddings=documents_embedding.tolist(),
    metadatas=metadatas
)
collections.count()
```

```python
collections.get()
```

```python
query="I enjoy Coding"
query_embedding=model.encode(query)
```

```python
ans=collections.query(
    query_embeddings=[query_embedding.tolist()],
    n_results=2,
    where={"topic":"Backend"}
)
```
