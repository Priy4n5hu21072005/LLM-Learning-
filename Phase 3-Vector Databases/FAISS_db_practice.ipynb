```python
import faiss
index=faiss.IndexFlatL2(384)
index
```

```python
from sentence_transformers import SentenceTransformer 
model=SentenceTransformer("all-MiniLM-L6-v2")
```

```python
documents = [
    "I love programming.",
    "I really enjoy coding.",
    "Programming is my favorite thing."
]
documents_embeddings=model.encode(documents)
documents_embeddings.shape
```

```python
index.add(documents_embeddings)
index.ntotal
```

```python
metadata = [
    {"topic": "Machine Learning", "language": "python"},
    {"topic": "Backend", "language": "java"},
    {"topic": "Programming", "language": "python"}
]

print(metadata)
```

```python
def sementic_search(query,k):
    query_embedding=model.encode([query])
    distances,indices=index.search(query_embedding,k)
    ans=[]
    for i in range(k):
        document=documents[indices[0][i]]
        distance=distances[0][i]
        meta=metadata[indices[0][i]]
        ans.append({
            "document":document,
            "distance":distance,
            "meta":meta
            })
    return ans
```

```python
results = sementic_search("I enjoy coding", 2)

for result in results:
    print(result)
```
