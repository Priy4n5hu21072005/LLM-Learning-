```python
from transformers import AutoTokenizer
```

```python
model_name = "Qwen/Qwen2.5-0.5B-Instruct"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
print(type(tokenizer))
```

```python
text = "Unbeleivalble"
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
text = "I Love you"
tokens = tokenizer.tokenize(text)
tokensId = tokenizer.encode(text)
print(tokens)
print(tokensId)
```

```python
decode = tokenizer.decode(tokensId)
print(decode)
```
