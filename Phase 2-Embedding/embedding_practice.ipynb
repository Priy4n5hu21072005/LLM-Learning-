```python
documents = [
    "Python is a programming language widely used for machine learning and data science.",
    "Spring Boot is a Java framework used to build backend applications.",
    "Football is a popular sport played around the world."
]

documents_embedding=model.encode(documents)
```

```python
query="which langguage is used for machine learning ?"
query_embedding=model.encode(query)
query_embedding.shape
```

```python
from sentence_transformers import util
documents_similarity=util.cos_sim(
    documents_embedding,
    query_embedding
)
documents_similarity
```

```python
def sementic_search(query,documents,documents_embedding):
    query_embedding=model.encode(query)
    similarity=util.cos_sim(
        query_embedding,
        documents_embedding
    )[0]
    best_index=similarity.argmax()
    return documents[best_index],similarity[best_index]
```

```python
result, score = sementic_search(
    "How can I develop backend applications?",
    documents,
    documents_embedding
)

print("Document:", result)
print("Score:", score)
```
